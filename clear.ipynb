## Valores Nulos

```python
print(df.isnull().sum())
```

## Que hacer con los nulos 

```python
print(df[df["Nombre"].isnull()])
```

```python
print(df[df["Ventas"].isnull()])
```

## Limpiar Ventas

```python
df["Ventas"] = (
    df["Ventas"]
    .astype(str)
    .str.replace(",", "", regex=False)
)
df["Ventas"] = pd.to_numeric(df["Ventas"], errors="coerce")

print(df["Ventas"].dtype)
```

## Limpiando Texto

```python
df["Nombre"] = df["Nombre"].str.strip()
#Pepe
```

## Ciudades Inconsistentes

Bogotá -> bogotà
bogota -> bogota
BOGOTA
Bogota
 Bogotá

```python
df["Ciudad"] = df["Ciudad"].str.lower()
```

```python
ciudades = {
    "bogota" : "Bogotá",
    "bogotá" : "Bogotá",
    "medellin" : "Medellín",
    "medellín" : "Medellín",
    "cali" : "Cali",
    "barranquilla" : "Barranquilla",
    "Villavicencio" : "Villavicencio"
}
```

## Categorias

```python
print(df["Categoria"].value_counts(dropna=False))
```

```python
df["Categoria"] = df["Categoria"].str.lower()
```

```python
categorias = {
    "tecnologia" : "Tecnología",
    "hogar" : "Hogar",
    "moda" : "Moda",
    "deportes" : "Deportes"
}
df["Categoria"] = df["Categoria"].map(categorias)
```

```python
print(df["Categoria"].value_counts(dropna=False))
```

## Edades

```python
print(df["Edad"].describe())
```

```python
df.loc[(df["Edad"] < 18) | (df["Edad"]> 100), "Edad"] = None
```

```python
print(df["Edad"].isnull().sum())
```

## Fechas

```python
print(df["Fecha"])
```

```python
df["Fecha"] = pd.to_datetime(
    df["Fecha"],
    errors="coerce",
    dayfirst=True
)
```

```python
print(df["Fecha"].dtype)
```

```python
print(df["Fecha"].min)
print(df["Fecha"].max)
```

## Duplicados

```python
print(df.duplicated().sum())
```

```python
print(df[df.duplicated()])
```

```python
df = df.drop_duplicates()
```

## Valores Negativos

```python
print(df[df["Ventas"] < 0])
```

```python
df.loc[df["Ventas"] < 0, "Ventas"] = None
```

## Borras los Nulos

```python
print(df.isnull().sum())
```

```python
df["Edad"] = df["Edad"].fillna(df["Edad"].median())
```

```python
df["Ventas"] = df["Ventas"].fillna(df["Ventas"].median())
```

```python
df = df.dropna(subset=["Nombre", "Ciudad", "Categoria"])
```
